## **Script 1:** Glasser+Tian Atlas Extraction

```python
"""
Glasser+Tian Atlas Extraction Script

Extract time courses from the 414-region Glasser+Tian atlas
from existing LCMV source estimates.
"""

import numpy as np
import pandas as pd
from lcmv_xtra import gt_extraction

# ======================
# CONFIGURATION
# ======================

# Path to your subject's output directory 
SUBJECT_OUTPUT_DIR = 'path_input' # folder where the source_estimate_LCMV.h5 is located

# Path to your fsaverage directory (created by download_fsaverage)
FS_DIR = 'path/_fs'

# Output filename prefix (optional - will use subject directory if not specified)
OUTPUT_PREFIX = None  # Set to None to use subject directory, or specify custom path

# ======================
# EXTRACTION
# ======================

print("🚀 Starting Glasser+Tian Atlas Extraction")
print(f"Subject output dir: {SUBJECT_OUTPUT_DIR}")
print(f"fsaverage directory: {FS_DIR}")

# Run extraction
gt_tc, roi_names = gt_extraction(
    subject_output_dir=SUBJECT_OUTPUT_DIR,
    global_subjects_dir=FS_DIR,
    verbose=True
)

print(f"✅ Extraction completed!")
print(f"Time courses shape: {gt_tc.shape}")
print(f"Number of ROIs: {len(roi_names)}")

# ======================
# VALIDATION
# ======================

print("\n📊 Data Validation:")
print(f"Data type: {gt_tc.dtype}")
print(f"Min value: {gt_tc.min():.6f}")
print(f"Max value: {gt_tc.max():.6f}")
print(f"Mean value: {gt_tc.mean():.2e}")
print(f"First 5 ROI names: {roi_names[:5]}")

# ======================
# SAVE RESULTS
# ======================

if OUTPUT_PREFIX is None:
    output_prefix = f"{SUBJECT_OUTPUT_DIR}/sub-01_task"
else:
    output_prefix = OUTPUT_PREFIX

# Save time courses
np.save(f"{output_prefix}_gt_timecourses.npy", gt_tc)
print(f"💾 Saved time courses: {output_prefix}_gt_timecourses.npy")

# Save ROI names
roi_df = pd.DataFrame({'roi_name': roi_names})
roi_df.to_csv(f"{output_prefix}_gt_roi_names.csv", index=False)
print(f"💾 Saved ROI names: {output_prefix}_gt_roi_names.csv")

# Create a combined DataFrame for easy analysis
gt_df = pd.DataFrame(gt_tc.T, columns=roi_names)
gt_df.to_csv(f"{output_prefix}_gt_timecourses.csv", index=False)
print(f"💾 Saved CSV format: {output_prefix}_gt_timecourses.csv")

print(f"\n🎉 All files saved successfully!")
print(f"Output directory: {SUBJECT_OUTPUT_DIR}")
```

## **Script 2:** DiFuMo Atlas Extraction

```python
"""
DiFuMo Atlas Extraction Script

Extract time courses from the 512-component DiFuMo atlas
from existing LCMV source estimates.
"""

import numpy as np
import pandas as pd
from lcmv_xtra import difumo_extraction

# ======================
# CONFIGURATION
# ======================

# Path to your subject's output directory 
SUBJECT_OUTPUT_DIR = 'path_input' # folder where the source_estimate_LCMV.h5 is located

# Path to your fsaverage directory (created by download_fsaverage)
FS_DIR = 'path/_fs'

# DiFuMo configuration (default: 512 components, 2mm resolution)
DIFUMO_CONFIG = {
    'dimension': 512,      # Options: 64, 128, 256, 512, 1024
    'resolution_mm': 2     # Options: 2 or 3
}

# Output filename prefix (optional - will use subject directory if not specified)
OUTPUT_PREFIX = None  # Set to None to use subject directory, or specify custom path

# ======================
# EXTRACTION
# ======================

print("🚀 Starting DiFuMo Atlas Extraction")
print(f"Subject output dir: {SUBJECT_OUTPUT_DIR}")
print(f"fsaverage directory: {FS_DIR}")
print(f"DiFuMo config: {DIFUMO_CONFIG['dimension']} components, {DIFUMO_CONFIG['resolution_mm']}mm")

# Run extraction
difumo_tc, component_info = difumo_extraction(
    subject_output_dir=SUBJECT_OUTPUT_DIR,
    global_subjects_dir=FS_DIR,
    difumo_config=DIFUMO_CONFIG,
    verbose=True
)

print(f"✅ Extraction completed!")
print(f"Time courses shape: {difumo_tc.shape}")
print(f"Number of components: {len(component_info)}")

# ======================
# VALIDATION
# ======================

print("\n📊 Data Validation:")
print(f"Data type: {difumo_tc.dtype}")
print(f"Min value: {difumo_tc.min():.6f}")
print(f"Max value: {difumo_tc.max():.6f}")
print(f"Mean value: {difumo_tc.mean():.2e}")

# Count components with assigned sources
valid_components = sum(1 for info in component_info if info['n_sources'] > 0)
print(f"Components with sources: {valid_components}/{len(component_info)}")

# ======================
# SAVE RESULTS
# ======================

if OUTPUT_PREFIX is None:
    output_prefix = f"{SUBJECT_OUTPUT_DIR}/sub-01_task"
else:
    output_prefix = OUTPUT_PREFIX

# Save time courses
np.save(f"{output_prefix}_difumo_timecourses.npy", difumo_tc)
print(f"💾 Saved time courses: {output_prefix}_difumo_timecourses.npy")

# Save component information
component_df = pd.DataFrame(component_info)
component_df.to_csv(f"{output_prefix}_difumo_component_info.csv", index=False)
print(f"💾 Saved component info: {output_prefix}_difumo_component_info.csv")

# Create a combined DataFrame for easy analysis
difumo_df = pd.DataFrame(difumo_tc.T, columns=[f'component_{i}' for i in range(difumo_tc.shape[0])])
difumo_df.to_csv(f"{output_prefix}_difumo_timecourses.csv", index=False)
print(f"💾 Saved CSV format: {output_prefix}_difumo_timecourses.csv")

print(f"\n🎉 All files saved successfully!")
print(f"Output directory: {SUBJECT_OUTPUT_DIR}")
print(f"\n💡 Note: DiFuMo extraction may take several minutes due to its probabilistic nature.")
```
